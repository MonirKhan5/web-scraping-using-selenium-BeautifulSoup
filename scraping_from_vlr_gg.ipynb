```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
```

```python
with open('vlr_gg.html','r') as f:
  html = f.read()
```

```python
soup = BeautifulSoup(html,'lxml')
```

```python
data = []
```

```python
for i in soup.find_all('tr')[1:]:
  names = i.find('div',class_='st-pl-name text-of')
  teams = i.find('div',class_='st-pl-country')
  rounds = i.find('td',attrs={'data-col':'rnd'})
  ratings = i.find('td',attrs={'data-col':'rating2'})
  combat = i.find('td',attrs={'data-col':'acs'})
  kd = i.find('td',attrs={'data-col':'kd'})
  kast = i.find('td',attrs={'data-col':'kast'})
  avg_damage = i.find('td',attrs={'data-col':'adr'})
  kpr = i.find('td',attrs={'data-col':'kpr'})
  apr = i.find('td',attrs={'data-col':'apr'})
  fkfd = i.find('td',attrs={'data-col':'fkfd'})
  hs = i.find('td',attrs={'data-col':'hsp'})
  cl_won = i.find('span',class_='st-cl-x')
  cl_played = i.find('span',class_='st-cl-y')
  max_kill = i.find('td',class_='mod-a mod-kmax')
  k = i.find('td',attrs={'data-col':'k'})
  d = i.find('td',attrs={'data-col':'d'})
  a = i.find('td',attrs={'data-col':'a'})

  data.append({
      'Players':names.text if names else np.nan,
      'Team':teams.text if teams else np.nan,
      'Rounds_Played':rounds.text if rounds else np.nan,
      'Rating':ratings.text if ratings else np.nan,
      'Avg_Combat_Score':combat.text if combat else np.nan,
      'KD':kd.text if kd else np.nan,
      'KAST':kast.text if kast else np.nan,
      'Avg_Damage_Per_Round':avg_damage.text if avg_damage else np.nan,
      'Kills_Per_Round':kpr.text if kpr else np.nan,
      'Assists_Per_Round':apr.text if apr else np.nan,
      'FKFD':fkfd.text if fkfd else np.nan,
      'Headshot_Percentage':hs.text if hs else np.nan,
      'Clutches_Played':cl_played.text if cl_played else np.nan,
      'Clutches_Won':cl_won.text if cl_won else np.nan,
      'Max_Kills':max_kill.text if max_kill else np.nan,
      'Total_Kills':k.text if k else np.nan,
      'Total_Deaths':d.text if d else np.nan,
      'Total_Assists':a.text if a else np.nan
  })
```

```python
valorant = pd.DataFrame(data)
```

```python
valorant.to_csv('valorant_data.csv')
```
